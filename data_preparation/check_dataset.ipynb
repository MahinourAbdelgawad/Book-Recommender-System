```python
import duckdb

# take a peek a the original dataset to see what the data looks like

con = duckdb.connect()
con.sql("SET memory_limit='6GB'")

print(con.sql("SELECT * FROM read_csv_auto('../dataset_original/goodreads_interactions.csv') LIMIT 10"))
```

```python
print(con.sql("SELECT * FROM read_csv_auto('../dataset_original/book_id_map.csv') LIMIT 10"))
```

```python
print(con.sql("""
    DESCRIBE SELECT * FROM read_json_auto(
        '../dataset_original/goodreads_book_authors.json',
        format='newline_delimited',
        maximum_object_size=20000000
    )
"""))
```

```python
print(con.sql("""
    DESCRIBE SELECT * FROM read_json_auto(
        '../dataset_original/goodreads_book_genres_initial.json',
        format='newline_delimited',
        maximum_object_size=20000000
    )
"""))
```

```python
con.sql("""
    SELECT column_name, column_type
    FROM (DESCRIBE SELECT * FROM read_json_auto(
        '../dataset_original/goodreads_books.json',
        format='newline_delimited',
        maximum_object_size=20000000))
""").show(max_rows=50)
```
